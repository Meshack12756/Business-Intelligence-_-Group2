```python
import os
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv(r"C:\Users\Admin\Desktop\Amazon\amazon_sales_data_2025.csv")

#This shows the basic information we have on the dataset we obtained
print("This is the basic information on the dataset:")
df.info()

#Display the shape of the data we have
print("Shape of the data:")
display(df.shape)

#Display the preview of the data
print("Preview of the data:")
display(df.head)

#Display the unique values count 
print("Unique values count")
display(df.nunique)

#This is the detected categorical column
categorical_cols = df.select_dtypes(include=['object']).columns
print("Categorical columns")
print(list(categorical_cols))
```

```python
#Removal of any duplicates in the dataset
df = df.drop_duplicates()
print("Duplicates removed. New shape:", df.shape)
```

```python
#The numerical columns from what we have
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
```

```python
#Skewness and Kurtosis
print("Below is the skewness and kurtosis of the data we have: ")
for col in numeric_cols:
    print(f"{col} -> Skew: {df[col].skew():.3f}, Kurtosis: {df[col].kurt():.3f}")
```

```python
#Removal of z-score outliers
df_numeric = df[numeric_cols]
```

```python
#Calculation of the z-score
z_scores = np.abs(stats.zscore(df_numeric, nan_policy='omit'))

#Rows with any z-score greater than 3 are outliers
outliers = (z_scores > 3).any(axis=1)
print(f"\nOutliers removed: {outliers.sum()} rows")

#Only non-outliers are kept
df = df[~outliers]
print("New shape after removing outliers:", df.shape)
```

```python
#Mapping of the data
if "status" in df.columns:
    mapping = {"low": 1, "medium": 2, "high": 3}
    df["status_mapped"] = df["status"].map(mapping)
```

```python
#Label encoding of the data
label_df = df.copy()

for col in categorical_cols:
    le = LabelEncoder()
    label_df[col] = le.fit_transform(label_df[col].astype(str))

print("Label Encoding complete.")
```

```python
#One-hot encoding of the data
onehot_df = pd.get_dummies(df, columns=categorical_cols, drop_first=False)
print("One-hot encoding complete.")
```

```python
df.to_csv(r"C:\Users\Admin\Desktop\Amazon\cleaned_Amazon_data.csv", index=False)
label_df.to_csv(r"C:\Users\Admin\Desktop\Amazon\label_encoded_data.csv", index=False)
onehot_df.to_csv(r"C:\Users\Admin\Desktop\Amazon\onehot_encoded_data.csv", index=False)

print("Files saved successfully.")
```
